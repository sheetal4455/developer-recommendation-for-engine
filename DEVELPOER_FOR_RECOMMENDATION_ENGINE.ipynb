```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score

df = pd.read_csv("/content/spotify_millsongdata.csv")
df = df.dropna(subset=['text', 'artist', 'song'])
df = df.drop_duplicates(subset=['artist', 'song'])
top_artists = df['artist'].value_counts().head(10).index
df = df[df['artist'].isin(top_artists)]


X = df['text']
y = df['artist']


tfidf = TfidfVectorizer(stop_words='english', max_df=0.8, min_df=5, max_features=5000)
X_tfidf = tfidf.fit_transform(X)


X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, stratify=y, random_state=42)


logreg = LogisticRegression(max_iter=500, solver='saga', multi_class='multinomial')
logreg.fit(X_train, y_train)
y_pred_logreg = logreg.predict(X_test)


rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
def evaluate_model(y_true, y_pred, model_name):
    print(f"\n--- {model_name} ---")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred, average='weighted', zero_division=0))
    print("Recall:", recall_score(y_true, y_pred, average='weighted', zero_division=0))
    print("F1 Score:", f1_score(y_true, y_pred, average='weighted', zero_division=0))
    print("\nClassification Report:\n", classification_report(y_true, y_pred, zero_division=0))

evaluate_model(y_test, y_pred_logreg, "Logistic Regression")
evaluate_model(y_test, y_pred_rf, "Random Forest")
```
